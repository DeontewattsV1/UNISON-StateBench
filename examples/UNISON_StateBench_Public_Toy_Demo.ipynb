# UNISON-StateBench: public synthetic notebook demonstration

This is a **public**, deterministic demonstration of the MIT-licensed [UNISON-StateBench project](https://github.com/DeontewattsV1/UNISON-StateBench). It uses **12 authored synthetic cases** only and does **not** evaluate Gemma 4, real coding agents, or code patches.

`Unknown` is not silently converted to `False`; provenance, epistemic status, and authority are separate. No private 64-case pilot inputs, sealed reference answers, original research archives, access tokens, or private notebook data are included.


```python
# Public reproducibility demo. Not Gemma 4 inference or code repair.
import json, hashlib
cases=[
 ('D01', True,'verified','owner',True),('D02',True,'verified','owner',True),
 ('D03', True,'verified','owner',True),('D04',True,'verified','owner',True),
 ('D05',False,'verified','owner',False),('D06',None,'unknown','none',False),
 ('D07',None,'unknown','none',False),('D08',True,'claimed','assistant',False),
 ('D09',True,'claimed','tool',False),('D10',None,'claimed','tool',False),
 ('D11',True,'unknown','none',False),('D12',False,'claimed','assistant',False)
]
def guard(r):
    _,value,epistemic,authority,_ = r
    return value is True and epistemic=='verified' and authority=='owner'
results=[guard(c) for c in cases]
assert len(cases)==12 and len({c[0] for c in cases})==12
assert all(allowed==c[4] for allowed,c in zip(results,cases))
assert sum(results)==4 and len(cases)-sum(results)==8
roundtripped=json.loads(json.dumps(cases))
assert roundtripped[5][1] is None and roundtripped[4][1] is False
print(json.dumps({'evaluation':'DETERMINISTIC TOY ONLY',
  'synthetic_cases':12,'synthetic_allowed':4,'synthetic_blocked':8,
  'gemma4_inference_calls':0,'real_patch_tests':0},indent=2))
print('SHA256:',hashlib.sha256(json.dumps(cases,sort_keys=True).encode()).hexdigest())
```

## Interpretation and limitations

The guard agrees with all **authored** labels by construction. This is a unit-test demonstration, not an empirical accuracy estimate, a comparative agent benchmark, or scientific validation. The baseline is merely a hypothetical unsafe always-write canary; it is not a capable coding agent. The demonstration neither proves that the guard improves Gemma 4 nor measures false blocks on real authorized repository tasks.
